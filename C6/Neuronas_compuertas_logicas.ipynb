```python

# Funcion OR con una neurona

x1 = 0
x2 = 1
bias = 1

for x1 in [0, 1]:
    for x2 in [0, 1]:

        x = [x1, x2]

        w = [0.5, 1, 1]

        y = bias * w[0] + x[0] * w[1] + x[1] * w[2]

        print(f"Entradas: {x1}, {x2}")
        
        # funcion de activacion
        if y > 1:
            print("Salida: 1")
        else:
            print("Salida: 0")
    
```

```python
# Funcion AND con una neurona

x1 = 0
x2 = 0
bias = 1

for x1 in [0, 1]:
    for x2 in [0, 1]:

        x = [x1, x2]

        w = [-1.5, 1.5, 1.5]

        y = bias * w[0] + x[0] * w[1] + x[1] * w[2]

        print(f"Entradas: {x1}, {x2}")
        
        # funcion de activacion
        if y > 1:
            print("Salida: 1")
        else:
            print("Salida: 0")
```

```python
import numpy as np
import os

# Funcion OR entrenamiento con una neurona   
x1 = 0
x2 = 0  
bias = 1

# Paso 1. Inicializar los pesos.

w = np.random.rand(3)

# Paso 2. Calcular la salida de la neurona para cada combinación de entradas.

entradas = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
salida_deseada = np.array([[0], [1], [1], [1]])

bias = 1
funcion_de_activacion = 0

epocas = 4

for epoca in range(epocas):

    for i in range(len(entradas)):
        x0 = bias
        x1 = entradas[i][0]
        x2 = entradas[i][1]
        x = [x0, x1, x2]

        # imprimir los pesos y las entradas
        print(f"Pesos: {w}")

        # Calcular la salida de la neurona
        y = np.dot(x, w.T)

        print(f"Entradas: {x1}, {x2}")

        print(f"Salida: {y}")
            
        # funcion de activacion
        if y > funcion_de_activacion:
            y = 1
        else:
            y = 0

        print(f"Salida funcion de activacion: {y}")
        
        error = salida_deseada[i] - y    
        
        # Paso 3. Actualizar los pesos utilizando el algoritmo de aprendizaje.
        
        w = w + (error * x)
        
        print(f"Error: {error}")
        print(f"Pesos actualizados: {w}")
        print("--------------------------------------------------")

WOR = w
    
```
