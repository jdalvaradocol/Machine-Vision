```python
import numpy as np
import matplotlib.pyplot as plt

def distancia(punto, centro):
    return np.sqrt(np.sum((punto - centro)**2))

def distancia_centroide(puntos, centroides):

    clusters = []
    # Asignar cada punto al centro más cercano
    for punto in puntos:

        distancias = []

        for centro in centroides:
            d = distancia(punto, centro)
            distancias.append(d)
        
        cluster = np.argmin(distancias)
        clusters.append(cluster)

    clusters = np.array(clusters)
    
    return clusters

def recalcular_centroides(puntos, clusters, k):
    nuevos_centros = []
    for i in range(k):
        cluster_puntos = puntos[clusters == i]
        nuevo_centro = np.mean(cluster_puntos, axis=0)
        nuevos_centros.append(nuevo_centro)
    
    return np.array(nuevos_centros)

def graficar(puntos, centroides, clusters):
    # Graficar
    plt.figure(figsize=(8,6))

    colores = ['green', 'blue', 'orange', 'purple', 'cyan', 'magenta', 'yellow', 'brown', 'pink', 'gray']
        
    # Graficar puntos del dataset
    for i in range(clusters.shape[0]-1):
        plt.scatter(puntos[i, 0], puntos[i, 1], color=colores[clusters[i]])
        
    # Graficar centroides
    plt.scatter(centroides[:, 0], centroides[:, 1], color='red', marker='X', s=200, label='Centroides')

    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title("Puntos aleatorios y centroides iniciales")
    plt.legend()
    plt.show()
```

```python


# Generar 150 números aleatorios entre 0 y 100
x = np.random.uniform(0, 100, 150)
y = np.random.uniform(0, 100, 150)
z = np.random.uniform(0, 100, 150)

# Crear matriz de puntos (150 x 3)
puntos = np.column_stack((x, y, z))

plt.figure(figsize=(8, 6))
plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("150 puntos aleatorios")
plt.show()
```

```python
# Número de clusters
k = 3

# Generar posiciones iniciales aleatorias para los centroides
xk = np.random.uniform(0, 100, k)
yk = np.random.uniform(0, 100, k)

# Crear matriz de centros (k x 2)
centros = np.column_stack((xk, yk))

# Graficar
plt.figure(figsize=(8,6))

# Graficar puntos del dataset
plt.scatter(x, y, label="Datos")

# Graficar centroides
plt.scatter(xk, yk, color='red', marker='X', s=200, label='Centroides')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Puntos aleatorios y centroides iniciales")
plt.legend()

plt.show()
```

```python

# Calcular la distancia de cada punto a cada centroide 
# y asignar el cluster correspondiente

iteraciones = 10

for i in range(iteraciones):
    clusters = distancia_centroide(puntos, centros) 
    centros_anteriores = centros.copy()
   
    centros = recalcular_centroides(puntos, clusters, k)
    centros_nuevos = centros.copy()
    
    variacion = np.linalg.norm(centros_nuevos - centros_anteriores)    
    
    print(f"Iteración {i+1}")
    print(f"Variación: {variacion}")
    graficar(puntos, centros, clusters)
```
