```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Generación de datos aleatorios
np.random.seed(0)

grupo1 = np.random.randn(50,2) + [2,2]
grupo2 = np.random.randn(50,2) + [8,3]
grupo3 = np.random.randn(50,2) + [5,8]

X = np.vstack((grupo1, grupo2, grupo3))
plt.scatter(X[:,0], X[:,1])
plt.title("Datos Aleatorios")
plt.xlabel("X")
plt.ylabel("Y")     

```

```python
kmeans = KMeans(n_clusters=6)
kmeans.fit(X)

labels = kmeans.labels_
centroides = kmeans.cluster_centers_

plt.scatter(X[:,0], X[:,1], c=labels, cmap='viridis')
plt.scatter(centroides[:,0], centroides[:,1], c='red', marker='X', s=200)
plt.title("K-Means Clustering")
plt.xlabel("X") 
```

```python
inercia = []

K = range(1,10)

for k in K:
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X)
    inercia.append(kmeans.inertia_)

plt.plot(K, inercia, marker='o')
plt.xlabel("Número de clusters")
plt.ylabel("Inercia")
plt.title("Método del Codo")
plt.show()
```
